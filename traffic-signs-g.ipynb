```python
!pip install tf2onnx
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2, ResNet50, VGG19, EfficientNetB0
from sklearn.metrics import confusion_matrix, classification_report
import tf2onnx

# -------------------------------------------------------------------------
# 1. SETUP & CONFIGURATION
# -------------------------------------------------------------------------
# Update this path to match your uploaded Kaggle dataset directory
DATASET_DIR = "/kaggle/input/datasets/monayamhossain/traffic-signs/Train" 

IMG_SIZE = (128, 128)
BATCH_SIZE = 32
NUM_CLASSES = 22
EPOCHS = 20

CLASS_NAMES = {
    0: 'No Road', 1: 'Not Sign', 2: 'Speed limit (20km/h)',
    3: 'Speed limit (30km/h)', 4: 'Speed limit (50km/h)', 5: 'Speed limit (60km/h)',
    6: 'Speed limit (80km/h)', 7: 'End of speed limit (80km/h)', 8: 'No passing',
    9: 'Stop', 10: 'Dangerous curve left', 11: 'Dangerous curve right',
    12: 'Double curve', 13: 'Turn right ahead', 14: 'Turn left ahead', 15: 'Ahead only',
    16: 'Go straight or right', 17: 'Go straight or left', 18: 'Keep right',
    19: 'Keep left', 20: 'End of no passing', 21: 'General caution'
}
labels_list = [CLASS_NAMES[i] for i in range(NUM_CLASSES)]

# -------------------------------------------------------------------------
# 2. DATA LOADING & AUGMENTATION PIPELINE
# -------------------------------------------------------------------------
print("Loading datasets...")
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='categorical' # Automatically handles One-Hot Encoding
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='categorical'
)

# Sequential Augmentation Layer (Applied to Training Data)
data_augmentation = models.Sequential([
    layers.RandomRotation(0.15),
    layers.RandomZoom(0.15),
    layers.RandomContrast(0.15),
    layers.RandomTranslation(height_factor=0.1, width_factor=0.1),
])

# Optimize performance with prefetching
def prepare_dataset(ds, augment=False):
    if augment:
        ds = ds.map(lambda x, y: (data_augmentation(x, training=True), y), 
                    num_parallel_calls=tf.data.AUTOTUNE)
    return ds.prefetch(buffer_size=tf.data.AUTOTUNE)

train_ds = prepare_dataset(train_ds, augment=True)
val_ds = prepare_dataset(val_ds, augment=False)

# -------------------------------------------------------------------------
# 3. HELPER FUNCTIONS FOR EVALUATION & EXPORT
# -------------------------------------------------------------------------
def plot_and_save_metrics(history, model_name):
    """Generates and saves the accuracy and loss curves."""
    acc = history.history['accuracy']
    val_acc = history.history['val_accuracy']
    loss = history.history['loss']
    val_loss = history.history['val_loss']
    epochs_range = range(len(acc))

    plt.figure(figsize=(14, 5))
    
    # Accuracy Curve
    plt.subplot(1, 2, 1)
    plt.plot(epochs_range, acc, label='Training Accuracy', color='blue')
    plt.plot(epochs_range, val_acc, label='Validation Accuracy', color='orange')
    plt.title(f'{model_name} - Training and Validation Accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend(loc='lower right')
    plt.grid(True)

    # Loss Curve
    plt.subplot(1, 2, 2)
    plt.plot(epochs_range, loss, label='Training Loss', color='blue')
    plt.plot(epochs_range, val_loss, label='Validation Loss', color='orange')
    plt.title(f'{model_name} - Training and Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend(loc='upper right')
    plt.grid(True)
    
    plt.tight_layout()
    filename = f"{model_name}_learning_curves.png"
    plt.savefig(filename, dpi=300)
    plt.show()
    print(f"[Saved] Learning curves saved as: {filename}")


def plot_and_save_confusion_matrix(model, val_ds, model_name):
    """Generates, plots, and saves a distinct Confusion Matrix plot."""
    print(f"Generating Confusion Matrix for {model_name}...")
    y_true = []
    y_pred = []
    
    for x, y in val_ds:
        preds = model.predict(x, verbose=0)
        y_true.extend(np.argmax(y.numpy(), axis=1))
        y_pred.extend(np.argmax(preds, axis=1))
        
    cm = confusion_matrix(y_true, y_pred)
    
    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=labels_list, yticklabels=labels_list)
    plt.title(f'Confusion Matrix - {model_name}')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.xticks(rotation=90)
    plt.yticks(rotation=0)
    plt.tight_layout()
    
    filename = f"{model_name}_confusion_matrix.png"
    plt.savefig(filename, dpi=300)
    plt.show()
    print(f"[Saved] Confusion Matrix saved as: {filename}")
    
    # Also print text report to console
    print(f"\n--- Classification Report for {model_name} ---")
    print(classification_report(y_true, y_pred, target_names=labels_list))


def export_to_onnx(model, model_name):
    """Converts the Keras model to ONNX format and saves it."""
    print(f"Converting {model_name} to ONNX format...")
    spec = (tf.TensorSpec((None, 128, 128, 3), tf.float32, name="input"),)
    output_path = f"{model_name}.onnx"
    
    model_proto, _ = tf2onnx.convert.from_keras(model, input_signature=spec, output_path=output_path)
    print(f"[Saved] ONNX model successfully saved as: {output_path}\n")

# -------------------------------------------------------------------------
# 4. TRAINING DICTIONARY SETUP (Updated to avoid pre-instantiating layers)
# -------------------------------------------------------------------------
MODELS_TO_TRAIN = {
    "MobileNetV2": {
        "base": MobileNetV2, 
        # Using a lambda function ensures the layer is created fresh inside the loop
        "preprocess": lambda x: tf.keras.layers.Rescaling(1./127.5, offset=-1)(x)
    },
    "ResNet50": {
        "base": ResNet50, 
        "preprocess": lambda x: tf.keras.applications.resnet50.preprocess_input(x)
    },
    "VGG19": {
        "base": VGG19, 
        "preprocess": lambda x: tf.keras.applications.vgg19.preprocess_input(x)
    },
    "EfficientNetB0": {
        "base": EfficientNetB0, 
        # EfficientNet has internal rescaling; pass the input through directly without extra layers
        "preprocess": lambda x: x 
    }
}

# -------------------------------------------------------------------------
# 5. EXECUTION LOOP
# -------------------------------------------------------------------------
for model_name, cfg in MODELS_TO_TRAIN.items():
    print("\n" + "="*60)
    print(f" STARTING TRAINING PIPELINE FOR: {model_name} ")
    print("="*60)
    
    # 1. Clear background sessions to reset the Keras layer naming counter
    tf.keras.backend.clear_session()
    
    # 2. Build Model Architecture
    inputs = layers.Input(shape=(128, 128, 3))
    
    # Apply preprocessing dynamically inside the running graph loop
    preprocessed_inputs = cfg["preprocess"](inputs)
    
    # Pass the preprocessed stream into the transfer learning backbone
    base_model = cfg["base"](weights='imagenet', include_top=False, input_tensor=preprocessed_inputs)
    base_model.trainable = False # Freeze base weights
    
    # Add custom head
    x = layers.GlobalAveragePooling2D()(base_model.output)
    x = layers.Dropout(0.3)(x)
    outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)
    
    model = models.Model(inputs, outputs)
    
    # 3. Compile Model
    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    
    # 4. Train Model
    history = model.fit(
        train_ds,
        validation_data=val_ds,
        epochs=EPOCHS,
        verbose=1
    )
    
    # 5. Generate Visual Artifacts
    print(f"\n## Evaluating Artifacts for {model_name} ##")
    plot_and_save_metrics(history, model_name)
    plot_and_save_confusion_matrix(model, val_ds, model_name)
    
    # 6. Export Model to ONNX
    export_to_onnx(model, model_name)

print("\n" + "="*60)
print(" ALL PIPELINES COMPLETED SUCCESSFULLY! ")
print("="*60)
print("You can download the artifacts from Kaggle's /kaggle/working directory output panel.")
print("="*60)
```
